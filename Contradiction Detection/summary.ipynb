# Time Complexity in Different Contradiction Detection 

1. Self Contradiction: $O(N)$

# No. of Calls in Different Contradiction Detection

It can be calculated using 
$${}^N C_K$$

1. Self Contradiction: Choose 1 from N
$${}^N C_1 = N$$

2. Pair Contradiction: Choose 2 from N
$${}^N C_2 = \frac{N(N-1)}{2}$$

3. Conditional Contradiction: Choose 3 from N
$${}^N C_3 = \frac{N(N-1)(N-2)}{6}$$

# Natural Language Inference (NLI)

A traditional approach, also called Recognizing Textual Entailment (RTE) where the task is to find the logical realtionship between two pieces of text (only 2, this is the limitation). 


The 2 pieces of text are called:
1. A **premise** (given statement)
2. A **hypothesis** (a statement to be evaluated against the premise)

- Solving a NLI requires Semantic Understanding, and Logical Reasoning (handling Negation, Quantifiers etc.)
- Key Datasets are: **SNLI, MultiNLI, ANLI** (Adversarial NLI)

Natural language inference (NLI) is one of the most important natural language understanding (NLU) tasks. NLI expresses the ability to infer information during spoken or written communication. The NLI task concerns the determination of the entailment relation of a pair of sentences, called the premise and hypothesis. If the premise entails the hypothesis, the pair is labeled as an “entailment”. If the hypothesis contradicts the premise, the pair is labeled a “contradiction”, and if there is not enough information to infer a relationship, the pair is labeled as “neutral”.

### The Three Labels

| Label | Meaning | Example |
|---|---|---|
| **Entailment** | The hypothesis logically follows from the premise | Premise: *"A man is eating pizza."* → Hypothesis: *"A person is eating food."* |
| **Contradiction** | The hypothesis is logically incompatible with the premise | Premise: *"A man is eating pizza."* → Hypothesis: *"No one is eating anything."*  |
| **Neutral** | Neither follows nor contradicts | Premise: *"A man is eating pizza."* → Hypothesis: *"The man is Italian."*  |

### Application of NLI

- Fact Verification (eg. does this article support or refuse a claim)
- Question Answering
- Summarization Evaluation
- Zero-shot text classification

# Data Generation Method in each contradiction

## 1. Self Contradiction: 
1. Take the original Statement
2. Synthetically flip the meaning
3. Hide it in a paragrpah
4. Add it back..
5. Now the document contains both Original Statement and Contradicting Statement 

## 2. Pair Contradiction
1. The procedure is same, but instead of adding to the same document, we add the synthetic statement to another document
2. There cna be 2 configurations, "near and far"
3. In near, we add to adjacent or nearby documents
4. In far, we add to distant documents. 

## 3. Conditional Contradiction
1. Start with one document, extract a topic sentence
2. Generate 3 new documents (d1,d2,d3), with special rules.
