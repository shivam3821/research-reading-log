# Context Validation
Context validation means checking whether the given information (documents/context) is:
1. consistent
2. non-contradictory
3. reliable

before using it to generate an answer.

### **In this paper,** context validation is defined as a process that:
1. Detects contradiction
2. Classify the type (self/pair/conditional)
3. Identifies sources (which documents are conflicting)

# Types of Conflicts
1. Context-Memory Conflicts
2. Context-Context Conflicts

### 1. **Context-Memory Conflicts**
Conflict between 
- RAG memory (context), and
- Model's internal knowledge (memory)

### 2. **Context-Context conflicts**
- Conflicts between RAG documents themselves
- Example:
    - **Doc1:** "India won the world cup"
    - **Doc2:** "Australia won the world cup"
    - Both document contradicts each other

> **Note:** In this paper we are focusing on Context-Context conflicts

- It arises due to
    - Outdated info
    - Comes from different source

# Why contradiction detection is challenging
### 1. Lack of dataset
Dataset not available that focuses on contradiction detection

### 2. Contradiction are complex
Some contradction can be as simple as, different news present in different documents. Some can be very complex that requires logical reasoning

### 3. Even human struggles
Research shows that, even human struggles when it comes to detecting contradictions

### 4. LLMs perform poorly
Even the state-of-the art LLMs (like ChatGPT, LLaMA), performs slightly better than random guessing

# Context Validators
Context validators are systems (or models) that check whether the given documents are consistent and non-contradictory before generating an answer.

> **Note:** Here LLMs are acting as Context Validators

# Types of conflicts
### 1. **Self Contradiction:**
Inside one document

### 2. **Contradicting Pair Documents:**
Between a pair of documents

### 3. **Conditional Contradiction:**
Between 3 documents, where info in one document creates contradiction between other 2 documents.
